```python
!pip install --upgrade snscrape
import snscrape.modules.twitter as sntwitter
import pandas as pd
import time

# Query untuk mencari tweet dengan kata-kata terkait depresi (Bahasa Indonesia)
query = "depresi OR bersalah OR stres OR cemas OR diabaikan OR dikucilkan OR frustasi OR gagal OR galau OR gelisah OR hampa OR kecewa OR kesepian OR menyerah OR sedih OR sendirian OR terasing OR terpuruk OR pasrah OR lelah lang:id since:2024-01-01 until:2025-01-01"
tweets = []

# Looping untuk mengambil tweet
for i, tweet in enumerate(sntwitter.TwitterSearchScraper(query).get_items()):
    try:
        if i >= 5000:  # Batasi jumlah tweet hingga 5000
            break
        if tweet is not None:  # Hindari error jika tweet kosong
            tweets.append([tweet.date, tweet.user.username, tweet.content])

        # Tunggu sebentar agar tidak terdeteksi sebagai spam
        time.sleep(2)  # Bisa diatur sesuai kebutuhan
    except Exception as e:
        print(f"Error saat scraping tweet ke-{i}: {e}")
        continue  # Lanjutkan ke iterasi berikutnya jika ada error

# Simpan hasil dalam DataFrame
df = pd.DataFrame(tweets, columns=['Date', 'User', 'Tweet'])

# Simpan ke Google Drive atau penyimpanan lokal
df.to_csv('/content/drive/MyDrive/SKRIPSI/tweet_depresi2.csv', index=False, encoding='utf-8')
```
